```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter  

sns.set(style="whitegrid")
plt.rcParams["font.size"] = 12

DF = pd.read_csv("./rsc_capsules_all_pages.csv")
DF = DF.dropna()
### 저널별 통계
journal_list = [str(x).split(",")[0].strip() for x in DF["journal_info"].fillna("").tolist()]
journal_count = Counter(journal_list)
## Counter → pandas DataFrame 변환
journal_df = (
    pd.DataFrame(journal_count.items(), columns=["journal", "count"])
    .sort_values("count", ascending=False)
    .reset_index(drop=True)
)
journal_df.head(10)
## 3) 시각화: 저널별 논문 수 (Bar Plot)
plt.figure(figsize=(10, 10))
sns.barplot(
    data=journal_df,
    y="journal",
    x="count",
    palette="viridis"
)
plt.title("Journal Article Count")
plt.xlabel("Count")
plt.ylabel("Journal")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

type_counts = DF["article_type"].value_counts()

plt.figure(figsize=(12, 8))
ax = sns.barplot(
    y=type_counts.index,
    x=type_counts.values,
    palette="viridis"
)

# 막대 끝에 숫자 표시
for i, v in enumerate(type_counts.values):
    ax.text(v + 0.5, i, str(v), va="center")

plt.title("Article Type Distribution", fontsize=16)
plt.xlabel("Count")
plt.ylabel("Article Type")
plt.tight_layout()
plt.show()

```
