```python
import json

json_path = "../section_identify/03_extracted_content.json"
json_data = json.load(open(json_path, "r", encoding="utf-8"))
```

```python
empty_keys = [k.split("-")[2] for k, v in json_data.items() if v == {}]
```

```python
empty_keys
import pandas as pd

df_results = pd.read_csv('./03_keyword_check_results.csv')
found_files = df_results[df_results['Found'] == True]['File'].apply(lambda x: x.split('/')[0]).unique().tolist()
 
intersection = set(empty_keys) & set(found_files)
only_in_empty = set(empty_keys) - set(found_files)
only_in_found = set(found_files) - set(empty_keys)

print(f"Intersection: {len(intersection)}")
print(f"Only in empty_keys: {len(only_in_empty)}")
print(f"Only in found_files: {len(only_in_found)}")
```
