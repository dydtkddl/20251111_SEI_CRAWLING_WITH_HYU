```python
import pandas as pd 
DF = pd.read_csv("./01_현황로그.csv")
DF
```

```python
# PII를 기준으로 그룹화하여 Status 열의 고유 값(성공/실패 등)의 개수를 셉니다.
# 이를 통해 각 논문(PII)별로 다운로드 시도 결과가 어떠했는지 한눈에 파악할 수 있습니다.

status_summary = DF.groupby('PII')['Status'].value_counts().unstack(fill_value=0)

# 총 시도 횟수도 같이 보고 싶다면 아래 주석 해제
# status_summary['Total'] = status_summary.sum(axis=1)

status_summary.to_csv("01_현황로그.summary.csv")

```
