```python
import json
from collections import Counter

file_path = r'.\01_run_out\tei_paragraph_sentence.jsonl'
section_list = []

with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        data = json.loads(line)
        # GROBID 결과물에서 섹션 정보는 주로 'section_title' 또는 'head' 키에 저장됩니다.
        # 'section' 키가 없을 경우를 대비해 여러 가능한 키를 확인합니다.
        section_name = data.get('section_title') or data.get('head') or data.get('section') or 'Unknown'
        section_list.append(str(section_name).strip())

# 섹션별 빈도수 계산 및 내림차순 정렬
section_counts = Counter(section_list).most_common()

if not section_counts:
    print("데이터를 찾을 수 없습니다. JSON 구조를 확인하세요.")
else:
    for section, count in section_counts:
        print(f"{section}: {count}")
```
